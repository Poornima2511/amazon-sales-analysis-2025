```python
df.loc[
    ~df["bought_in_last_month"].fillna("").str.contains("bought in past month"),
    "bought_in_last_month"
].value_counts()
```

```python
df["estimated_monthly_purchases"] = (
    df["bought_in_last_month"]
    .where(
        df["bought_in_last_month"]
        .fillna("")
        .str.contains("bought in past month")
    )
)
```

```python
df
```

```python
extracted = df["estimated_monthly_purchases"].str.extract(
    r'(\d+(?:\.\d+)?)([Kk]?)',
    expand=True
)

extracted.columns = ["number", "unit"]
```

```python
extracted
```

```python
df["estimated_monthly_purchases"] = (
    pd.to_numeric(extracted["number"]) *
    extracted["unit"].str.lower().map({"k": 1000}).fillna(1)
)
```
