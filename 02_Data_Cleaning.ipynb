# 1.Cleaning the price column

# Cleaning the price on variant column

```python
df["variant_price_clean"] = (
    df["Price_on_variant"]
    .str.extract(r'\$(\d+(?:\.\d+)?)')[0]
    .astype(float)
)
```

```python
df[["variant_price_clean","Price_on_variant"]].head(20)
```

# Cleaning the listed price

```python
df["Listed_price_clean"] = (
    df["Listed_price"]
    .str.extract(r'\$(\d+(?:\.\d+)?)')[0]
    .astype(float)
)
```

```python
df["Listed_price_clean"]
```

```python
df[["Current/discounted_price","variant_price_clean","Listed_price_clean"]].head(25)
```

```python
df["final_price"] = (
    df["Current/discounted_price"]
    .fillna(df["variant_price_clean"])
    .fillna(df["Listed_price_clean"])
)
```

```python
df["final_price"].head(25)
```

```python
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')
```

```python
df[["Current/discounted_price","variant_price_clean","Listed_price_clean","final_price"]].head(25)
```

# 2.Clean Rating column

```python
df["Rating_clean"] =(
    df["Rating"]
    .str.extract(r'(\d\.\d)')
    .astype(float)
)
```

```python
df["Rating_clean"]
```

# 3.Clean the number of review column

```python
df["Number_of_reviews_clean"] = (
    df["Number_of_reviews"]
    .str.replace(",","",regex = False)
    .astype("Int64")
)
```

```python
df[["Number_of_reviews", "Number_of_reviews_clean"]].head(10)
```

# 4.Clean the bought in last month column


```python
df[['bought_in_last_month']].value_counts().head(30)
```

```python
import re
import numpy as np

def clean_bought(value):
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    match = re.search(r'([\d,.]+)\s*([Kk]?)\+', value)
    
    if not match:
        return np.nan
    
    number = float(match.group(1).replace(',', ''))
    suffix = match.group(2).upper()
    
    if suffix == 'K':
        number *= 1000
    
    return number

df['bought_in_last_month_clean'] = df['bought_in_last_month'].apply(clean_bought)
```
